# How to run King of the Hill in Colab:
1. Generate a repository ZIP file by running `king_of_the_hill/zip_project.sh`.
2. Upload `social-agents.zip` to Google Colab instance (can drag and drop it).
3. If you don't provide an `king_of_the_hill/wandb_api_key.txt`, you will be prompted to authenticate your weights and biases (`wandb`) account.

```python
# Train an agent
if resume:
  !export PYTHONPATH=$(pwd) && python $file_name resume --run-id $resume_id 
else:
  if multi_model:
    !export PYTHONPATH=$(pwd) && python $file_name new --num-gpus $num_gpus --num-workers $num_workers --multi-model
  else: 
    !export PYTHONPATH=$(pwd) && python $file_name new --num-gpus $num_gpus --num-workers $num_workers 
print('Done')
```
